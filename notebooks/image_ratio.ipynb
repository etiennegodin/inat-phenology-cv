```python
from PIL import Image
from PIL import UnidentifiedImageError
from pathlib import Path
import seaborn as sns
import numpy as np
```

```python
dir_path = Path('/home/etienne/projects/inat-phenology-cv/data/images')
files = [f.name for f in dir_path.iterdir() if f.is_file()]
```

```python
ratios = []
for i, f in enumerate(files):
    try:
        x,y = Image.open(dir_path/f).size
    except UnidentifiedImageError as e:
        continue
    if x <= y:
        ratio = y/x
    else:
        ratio = x/y
    ratios.append(round(ratio,3))
```

```python
print(np.array(ratios).mean())
print(np.array(ratios).std())
print(np.array(ratios).max())
print(np.array(ratios).min())
```

```python
sns.histplot(ratios, log_scale=(False,True))
```
