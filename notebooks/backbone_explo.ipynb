```python
import torch
import torch.nn.functional as F

encoder = model.visual
encoder.eval()

img_size = encoder.image_size
dummy_input_shape = (3, *img_size)
dummy_input = torch.zeros(1, *dummy_input_shape)

output_dim = encoder(dummy_input).shape[-1]

embeddings = encoder(dummy_input)
embeddings = F.normalize(embeddings, dim=1)  # embeddings normalised per row (img)
```
